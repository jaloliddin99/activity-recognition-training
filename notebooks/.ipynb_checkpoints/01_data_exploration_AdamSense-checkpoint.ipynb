## Step 4: Identify Wrist Sensor Columns

```python
print("="*70)
print("IDENTIFYING WRIST SENSOR COLUMNS")
print("="*70)

# AdamSense has two devices: pocket (_p) and wrist (_w)
# We only need wrist data for our microcontroller

# Check for wrist columns
wrist_accel = ['Ax_w', 'Ay_w', 'Az_w']
wrist_gyro = ['Gx_w', 'Gy_w', 'Gz_w']
wrist_sensors = wrist_accel + wrist_gyro

# Verify all columns exist
missing_cols = [col for col in wrist_sensors if col not in df.columns]

if missing_cols:
    print(f"\n❌ ERROR: Missing wrist sensor columns: {missing_cols}")
    print(f"\nAvailable columns: {list(df.columns)}")
    raise ValueError("Required wrist sensor columns not found")

print("\n✓ All wrist sensor columns found!")
print("\nWrist Sensors (for microcontroller):")
print(f"  Accelerometer: {wrist_accel}")
print(f"  Gyroscope:     {wrist_gyro}")
print(f"  Total: 6 channels")

# Check for magnetometer (we won't use it, but good to know)
wrist_mag = ['Mx_w', 'My_w', 'Mz_w']
has_magnetometer = all(col in df.columns for col in wrist_mag)

if has_magnetometer:
    print(f"\n⚠️  Magnetometer data present: {wrist_mag}")
    print("   → Will NOT be used (microcontroller doesn't have magnetometer)")

# Check for timestamp
if 'Time_w' in df.columns:
    timestamp_col = 'Time_w'
    print(f"\n✓ Wrist timestamp column found: {timestamp_col}")
else:
    print(f"\n⚠️  Warning: 'Time_w' column not found")
    print(f"   Available time columns: {[col for col in df.columns if 'time' in col.lower()]}")
    timestamp_col = None
```

## Step 9: SAMPLING RATE CALCULATION - CRITICAL!

```python
print("="*70)
print("SAMPLING RATE CALCULATION FROM TIMESTAMPS")
print("="*70)

if timestamp_col is None or timestamp_col not in df.columns:
    print("\n❌ ERROR: Timestamp column not found!")
    print(f"   Expected: 'Time_w'")
    print(f"   Available: {list(df.columns)}")
    raise ValueError("Timestamp column 'Time_w' not found in dataset")

print(f"\n✓ Using timestamp column: {timestamp_col}")

# Calculate sampling rate for multiple users
print("\nCalculating sampling rate from timestamps...")
print("-"*70)

sampling_rates = []
users_to_test = df['User'].unique()[:10]  # Test first 10 users

print(f"\nTesting {len(users_to_test)} users:")
print("-"*70)

for user in users_to_test:
    # Get user data, sorted by timestamp
    user_data = df[df['User'] == user].sort_values(timestamp_col)
    
    if len(user_data) < 100:
        continue
    
    # Calculate time differences
    time_diffs = user_data[timestamp_col].diff().dropna()
    
    # Remove zeros and negative values
    time_diffs = time_diffs[time_diffs > 0]
    
    if len(time_diffs) < 10:
        continue
    
    # Remove outliers (large gaps between activities)
    Q1 = time_diffs.quantile(0.25)
    Q3 = time_diffs.quantile(0.75)
    IQR = Q3 - Q1
    clean_diffs = time_diffs[(time_diffs >= Q1 - 1.5*IQR) & (time_diffs <= Q3 + 1.5*IQR)]
    
    if len(clean_diffs) == 0:
        continue
    
    # Calculate average and median intervals
    avg_interval = clean_diffs.mean()
    median_interval = clean_diffs.median()
    
    # Determine sampling rate based on interval magnitude
    if avg_interval > 100:  # Likely milliseconds
        rate = 1000.0 / avg_interval
        unit = "ms"
    elif avg_interval > 1:  # Likely seconds  
        rate = 1.0 / avg_interval
        unit = "s"
    else:  # Very small values
        rate = 1.0 / avg_interval
        unit = "s"
    
    sampling_rates.append(rate)
    print(f"  User {user:>3d}: {avg_interval:>8.2f} {unit} → {rate:>6.2f} Hz")

print("-"*70)

if len(sampling_rates) == 0:
    print("\n❌ ERROR: Could not calculate sampling rate!")
    print("   Check timestamp data quality")
    raise ValueError("Failed to calculate sampling rate")

# Calculate statistics
mean_rate = np.mean(sampling_rates)
median_rate = np.median(sampling_rates)
std_rate = np.std(sampling_rates)
min_rate = np.min(sampling_rates)
max_rate = np.max(sampling_rates)

print("\n" + "="*70)
print("SAMPLING RATE STATISTICS")
print("="*70)
print(f"\nBased on {len(sampling_rates)} users:")
print(f"  Mean:   {mean_rate:>6.2f} Hz")
print(f"  Median: {median_rate:>6.2f} Hz")
print(f"  Std:    {std_rate:>6.2f} Hz")
print(f"  Min:    {min_rate:>6.2f} Hz")
print(f"  Max:    {max_rate:>6.2f} Hz")

# Final estimate (use median for robustness)
final_sampling_rate = median_rate

print("\n" + "="*70)
print(f"🎯 FINAL SAMPLING RATE: {final_sampling_rate:.1f} Hz")
print("="*70)

# Check if it's a common rate
common_rates = [20, 25, 50, 100, 200]
closest_rate = min(common_rates, key=lambda x: abs(x - final_sampling_rate))

if abs(closest_rate - final_sampling_rate) < 5:
    print(f"\n✓ Close to standard rate: {closest_rate} Hz")
    final_sampling_rate = closest_rate
```

```python
# Calculate recommended window sizes
print("\n" + "="*70)
print("RECOMMENDED WINDOW SIZES")
print("="*70)

print(f"\nUsing sampling rate: {final_sampling_rate:.0f} Hz")
print("\nWindow calculations:")
print("-"*70)
print(f"{'Duration':>12} {'Samples':>12} {'Power of 2':>15} {'Recommended?':>15}")
print("-"*70)

durations = [2.0, 2.5, 3.0, 4.0, 5.0]
recommended_window = None

for duration in durations:
    samples = int(final_sampling_rate * duration)
    # Find nearest power of 2
    power_of_2 = 2 ** round(np.log2(samples))
    
    # Recommend 2.5-3 second windows
    is_recommended = '✓' if 2.5 <= duration <= 3.0 else ''
    
    print(f"{duration:>11.1f}s {samples:>12d} {power_of_2:>15d} {is_recommended:>15}")
    
    if 2.5 <= duration <= 3.0 and recommended_window is None:
        recommended_window = power_of_2

print("-"*70)

if recommended_window is None:
    recommended_window = 2 ** round(np.log2(final_sampling_rate * 2.5))

print(f"\n✅ RECOMMENDED WINDOW SIZE: {recommended_window} samples")
print(f"   → Duration: ~{recommended_window/final_sampling_rate:.2f} seconds")
print(f"   → Use this in your preprocessing CONFIG!")

print("\n" + "="*70)
```

## Step 11: Correlation Analysis

```python
print("="*70)
print("SENSOR CORRELATION ANALYSIS")
print("="*70)

# Calculate correlation matrix
corr_matrix = df[wrist_sensors].corr()

print("\nCorrelation matrix:")
display(corr_matrix)

# Visualize
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=1, cbar_kws={"shrink": 0.8},
            xticklabels=['Ax', 'Ay', 'Az', 'Gx', 'Gy', 'Gz'],
            yticklabels=['Ax', 'Ay', 'Az', 'Gx', 'Gy', 'Gz'])
plt.title('Wrist Sensor Correlation Matrix - AdamSense', fontsize=14, weight='bold', pad=20)
plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}correlation_matrix.png', dpi=300, bbox_inches='tight')
print(f"\n✓ Saved: {OUTPUT_DIR}correlation_matrix.png")
plt.show()

# Find highly correlated pairs
print("\nHighly correlated sensor pairs (|r| > 0.7):")
high_corr = []
for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        if abs(corr_matrix.iloc[i, j]) > 0.7:
            high_corr.append((corr_matrix.columns[i], corr_matrix.columns[j], corr_matrix.iloc[i, j]))

if high_corr:
    for sensor1, sensor2, corr_val in high_corr:
        print(f"  {sensor1} <-> {sensor2}: {corr_val:.3f}")
else:
    print("  None found (good - sensors are independent)")
```
