```python
"""
ADAM-Sense Dataset Exploration
Activity Recognition from Wrist IMU Sensors
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set plotting style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14, 6)

print("=" * 60)
print("ADAM-Sense Dataset Exploration")
print("=" * 60)

# ============================================================================
# STEP 1: Load Dataset
# ============================================================================
data_path = Path('../data/raw/dataset.csv')
df = pd.read_csv(data_path)

print(f"\n✓ Dataset loaded: {df.shape[0]:,} samples, {df.shape[1]} columns")

# ============================================================================
# STEP 2: Understand Data Structure
# ============================================================================
print("\n" + "=" * 60)
print("DATA STRUCTURE")
print("=" * 60)
print(f"\nColumn names:")
for i, col in enumerate(df.columns, 1):
    print(f"{i:2d}. {col}")

print(f"\nData types:")
print(df.dtypes)

print(f"\nMissing values:")
missing = df.isnull().sum()
if missing.sum() > 0:
    print(missing[missing > 0])
else:
    print("No missing values! ✓")

print(f"\nFirst 3 rows:")
print(df.head(3))

# ============================================================================
# STEP 3: Activity Distribution
# ============================================================================
print("\n" + "=" * 60)
print("ACTIVITY DISTRIBUTION")
print("=" * 60)

activity_counts = df['Activity'].value_counts()
print(f"\nTotal activities: {len(activity_counts)}")
print(f"\nSamples per activity:")
for activity, count in activity_counts.items():
    percentage = (count / len(df)) * 100
    print(f"  {activity:20s}: {count:6,} ({percentage:5.2f}%)")

# Plot activity distribution
plt.figure(figsize=(12, 5))
activity_counts.plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('Activity Distribution in ADAM-Sense Dataset', fontsize=14, weight='bold')
plt.xlabel('Activity', fontsize=12)
plt.ylabel('Number of Samples', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================================
# STEP 4: User Distribution
# ============================================================================
print("\n" + "=" * 60)
print("USER DISTRIBUTION")
print("=" * 60)

user_counts = df['User'].value_counts().sort_index()
print(f"\nTotal users: {len(user_counts)}")
print(f"\nSamples per user:")
for user, count in user_counts.items():
    print(f"  User {user}: {count:6,} samples")

# ============================================================================
# STEP 5: Sampling Rate Analysis
# ============================================================================
print("\n" + "=" * 60)
print("SAMPLING RATE ANALYSIS")
print("=" * 60)

# Analyze wrist timestamps
wrist_time = df['Time_w'].values
time_diffs = np.diff(wrist_time)

# Remove zeros and outliers
time_diffs = time_diffs[time_diffs > 0]
time_diffs = time_diffs[time_diffs < np.percentile(time_diffs, 99)]

avg_interval = np.mean(time_diffs)
median_interval = np.median(time_diffs)

# Determine if milliseconds or microseconds
if avg_interval > 1000:
    # Likely milliseconds
    sampling_rate = 1000 / avg_interval
    unit = "milliseconds"
elif avg_interval > 10:
    # Likely centiseconds (10ms units)
    sampling_rate = 100 / avg_interval
    unit = "centiseconds"
else:
    # Likely seconds or unknown
    sampling_rate = 1 / avg_interval
    unit = "seconds"

print(f"\nAverage time interval: {avg_interval:.2f} {unit}")
print(f"Median time interval: {median_interval:.2f} {unit}")
print(f"Estimated sampling rate: {sampling_rate:.2f} Hz")

# ============================================================================
# STEP 6: Sensor Data Statistics (WRIST sensors only)
# ============================================================================
print("\n" + "=" * 60)
print("WRIST SENSOR STATISTICS")
print("=" * 60)

wrist_sensors = ['Ax_w', 'Ay_w', 'Az_w', 'Gx_w', 'Gy_w', 'Gz_w', 
                 'Mx_w', 'My_w', 'Mz_w']

sensor_stats = df[wrist_sensors].describe()
print("\n", sensor_stats)

# ============================================================================
# STEP 7: Visualize Sensor Data for Each Activity
# ============================================================================
print("\n" + "=" * 60)
print("VISUALIZING SENSOR DATA")
print("=" * 60)

# Select a few representative activities
activities_to_plot = ['Nail_biting', 'Knuckles_cracking', 'Hand_tapping', 'Sitting']
available_activities = [act for act in activities_to_plot if act in df['Activity'].unique()]

if not available_activities:
    # Use first 4 activities if our selection isn't available
    available_activities = df['Activity'].unique()[:4]

print(f"\nPlotting activities: {', '.join(available_activities)}")

# Plot accelerometer data
fig, axes = plt.subplots(len(available_activities), 3, figsize=(15, 3*len(available_activities)))
if len(available_activities) == 1:
    axes = axes.reshape(1, -1)

accel_cols = ['Ax_w', 'Ay_w', 'Az_w']
accel_labels = ['Accel X', 'Accel Y', 'Accel Z']

for i, activity in enumerate(available_activities):
    activity_data = df[df['Activity'] == activity].head(200)  # First 200 samples
    
    for j, (col, label) in enumerate(zip(accel_cols, accel_labels)):
        axes[i, j].plot(activity_data[col].values, linewidth=0.8, color='darkblue')
        axes[i, j].set_ylabel(f'{label} (m/s²)')
        axes[i, j].grid(True, alpha=0.3)
        
        if i == 0:
            axes[i, j].set_title(label, fontsize=12, weight='bold')
        if i == len(available_activities) - 1:
            axes[i, j].set_xlabel('Sample')
        
        # Add activity label on left
        if j == 0:
            axes[i, j].text(-0.3, 0.5, activity, 
                          transform=axes[i, j].transAxes,
                          rotation=90, va='center', ha='center',
                          fontsize=11, weight='bold')

plt.suptitle('Wrist Accelerometer Data by Activity', fontsize=14, weight='bold', y=1.00)
plt.tight_layout()
plt.show()

# Plot gyroscope data
fig, axes = plt.subplots(len(available_activities), 3, figsize=(15, 3*len(available_activities)))
if len(available_activities) == 1:
    axes = axes.reshape(1, -1)

gyro_cols = ['Gx_w', 'Gy_w', 'Gz_w']
gyro_labels = ['Gyro X', 'Gyro Y', 'Gyro Z']

for i, activity in enumerate(available_activities):
    activity_data = df[df['Activity'] == activity].head(200)
    
    for j, (col, label) in enumerate(zip(gyro_cols, gyro_labels)):
        axes[i, j].plot(activity_data[col].values, linewidth=0.8, color='darkred')
        axes[i, j].set_ylabel(f'{label} (rad/s)')
        axes[i, j].grid(True, alpha=0.3)
        
        if i == 0:
            axes[i, j].set_title(label, fontsize=12, weight='bold')
        if i == len(available_activities) - 1:
            axes[i, j].set_xlabel('Sample')
        
        if j == 0:
            axes[i, j].text(-0.3, 0.5, activity, 
                          transform=axes[i, j].transAxes,
                          rotation=90, va='center', ha='center',
                          fontsize=11, weight='bold')

plt.suptitle('Wrist Gyroscope Data by Activity', fontsize=14, weight='bold', y=1.00)
plt.tight_layout()
plt.show()

# ============================================================================
# STEP 8: Correlation Analysis
# ============================================================================
print("\n" + "=" * 60)
print("SENSOR CORRELATION ANALYSIS")
print("=" * 60)

# Compute correlation for accelerometer and gyroscope only (skip magnetometer for now)
sensor_subset = ['Ax_w', 'Ay_w', 'Az_w', 'Gx_w', 'Gy_w', 'Gz_w']
correlation = df[sensor_subset].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Wrist Sensor Correlation Matrix (Accel + Gyro)', fontsize=14, weight='bold')
plt.tight_layout()
plt.show()

# ============================================================================
# SUMMARY & RECOMMENDATIONS
# ============================================================================
print("\n" + "=" * 60)
print("SUMMARY & RECOMMENDATIONS")
print("=" * 60)

print(f"""
Dataset Overview:
✓ Total samples: {len(df):,}
✓ Activities: {len(activity_counts)}
✓ Users: {len(user_counts)}
✓ Estimated sampling rate: {sampling_rate:.1f} Hz

Sensor Configuration:
✓ Wrist IMU sensors available:
  - Accelerometer (Ax_w, Ay_w, Az_w)
  - Gyroscope (Gx_w, Gy_w, Gz_w)
  - Magnetometer (Mx_w, My_w, Mz_w)

Recommendations for 8-week project:
1. Use WRIST sensors only (_w suffix) - matches your hardware
2. Start with 6 features: 3 accel + 3 gyro (skip magnetometer initially)
3. Select 4-5 most distinct activities for better accuracy:
   - Nail_biting
   - Knuckles_cracking
   - Hand_tapping
   - Sitting (as idle class)
   
4. Window size for {sampling_rate:.1f} Hz:
   - 2 seconds = {int(sampling_rate * 2)} samples
   - 3 seconds = {int(sampling_rate * 3)} samples
   - Recommended: {int(sampling_rate * 2.5)} samples (2.5 seconds)

5. Next steps:
   ✓ Data preprocessing (windowing, normalization)
   ✓ Train/test split by USER (not random) to avoid data leakage
   ✓ Build and train CNN model
""")

print("=" * 60)
print("Exploration complete! Ready for preprocessing.")
print("=" * 60)
```
