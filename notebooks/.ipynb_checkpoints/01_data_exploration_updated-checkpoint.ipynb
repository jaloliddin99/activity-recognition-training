## Step 5: Activity Distribution

```python
print("="*70)
print("ACTIVITY DISTRIBUTION")
print("="*70)

activity_counts = df['Activity'].value_counts()
total = len(df)

print(f"\nTotal activities: {len(activity_counts)}")
print("\nBreakdown:")
print("-"*70)
print(f"{'Activity':<30} {'Count':>12} {'Percentage':>12}")
print("-"*70)

for activity, count in activity_counts.items():
    pct = count / total * 100
    bar = '█' * int(pct / 2)
    print(f"{activity:<30} {count:>12,} {pct:>11.2f}% {bar}")

print("-"*70)
```

```python
# Visualize
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Bar plot
colors = sns.color_palette("husl", len(activity_counts))
activity_counts.plot(kind='bar', ax=axes[0], color=colors, edgecolor='black')
axes[0].set_title('Activity Distribution', fontsize=14, weight='bold')
axes[0].set_xlabel('Activity')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(True, alpha=0.3)

# Pie chart
axes[1].pie(activity_counts.values, labels=activity_counts.index,
           autopct='%1.1f%%', startangle=90, colors=colors)
axes[1].set_title('Activity Percentage', fontsize=14, weight='bold')

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}activity_distribution.png', dpi=300, bbox_inches='tight')
print(f"\n✓ Saved: {OUTPUT_DIR}activity_distribution.png")
plt.show()
```

## Step 6: User Distribution

```python
print("="*70)
print("USER DISTRIBUTION")
print("="*70)

user_counts = df['User'].value_counts().sort_index()

print(f"\nTotal users: {len(user_counts)}")
print(f"\nStatistics:")
print(f"  Min:    {user_counts.min():>8,} samples")
print(f"  Max:    {user_counts.max():>8,} samples")
print(f"  Mean:   {user_counts.mean():>8,.0f} samples")
print(f"  Median: {user_counts.median():>8,.0f} samples")

print(f"\nFirst 10 users:")
for user, count in user_counts.head(10).items():
    print(f"  User {user:>3d}: {count:>8,} samples")
```

## Step 7: Sensor Statistics

```python
print("="*70)
print("SENSOR STATISTICS")
print("="*70)

# Identify sensor columns based on dataset type
if DATASET_TYPE == 'StressSense':
    sensor_cols = ['Ax', 'Ay', 'Az', 'Gx', 'Gy', 'Gz']
    print("\n✓ StressSense: Using Accel (Ax,Ay,Az) + Gyro (Gx,Gy,Gz)")
    print("  Note: Magnetometer (Mx,My,Mz) present but NOT used in model")
elif DATASET_TYPE == 'AdamSense':
    # Check if wrist or pocket columns exist
    if 'Ax_w' in df.columns:
        sensor_cols = ['Ax_w', 'Ay_w', 'Az_w', 'Gx_w', 'Gy_w', 'Gz_w']
        print("\n✓ AdamSense: Using WRIST sensors (Accel + Gyro)")
    else:
        sensor_cols = ['Ax', 'Ay', 'Az', 'Gx', 'Gy', 'Gz']
        print("\n✓ AdamSense: Using standard sensor columns")
else:
    sensor_cols = ['Ax', 'Ay', 'Az', 'Gx', 'Gy', 'Gz']
    print("\n✓ Using standard sensor columns")

print(f"\nSensor columns: {sensor_cols}")

# Statistics
print("\nStatistics:")
print("-"*85)
print(f"{'Sensor':<10} {'Mean':>12} {'Std':>12} {'Min':>12} {'Max':>12} {'Range':>12}")
print("-"*85)

for sensor in sensor_cols:
    if sensor in df.columns:
        mean = df[sensor].mean()
        std = df[sensor].std()
        min_val = df[sensor].min()
        max_val = df[sensor].max()
        range_val = max_val - min_val
        print(f"{sensor:<10} {mean:>12.4f} {std:>12.4f} {min_val:>12.4f} {max_val:>12.4f} {range_val:>12.4f}")

# Missing values
print("\nMissing Values:")
missing = df[sensor_cols].isnull().sum()
if missing.sum() == 0:
    print("  ✓ No missing values!")
else:
    for sensor, count in missing.items():
        if count > 0:
            print(f"  {sensor}: {count:,} ({count/len(df)*100:.2f}%)")
```

---
## Step 8: Sampling Rate Estimation - CRITICAL!

```python
print("="*70)
print("SAMPLING RATE ESTIMATION")
print("="*70)

# Identify timestamp column
if DATASET_TYPE == 'AdamSense':
    if 'Time_w' in df.columns:
        timestamp_col = 'Time_w'
        print("\n✓ Using AdamSense wrist timestamp: Time_w")
    elif 'TimeStamp' in df.columns:
        timestamp_col = 'TimeStamp'
        print("\n✓ Using TimeStamp column")
    else:
        timestamp_col = None
        print("\n⚠️  No timestamp column found")
else:  # StressSense
    timestamp_col = 'TimeStamp'
    print("\n⚠️  StressSense timestamps are CORRUPTED (constant values)")
    print("   Will estimate from signal characteristics instead")

# Method 1: Try to calculate from timestamps (for AdamSense)
if timestamp_col and DATASET_TYPE == 'AdamSense':
    print("\n--- Method 1: Calculate from Timestamps ---")
    
    # Select a continuous segment from one user
    test_user = df['User'].iloc[0]
    user_data = df[df['User'] == test_user].sort_values(timestamp_col).head(1000)
    
    # Calculate time differences
    time_diffs = user_data[timestamp_col].diff().dropna()
    
    # Remove outliers (activity transitions, gaps)
    Q1 = time_diffs.quantile(0.25)
    Q3 = time_diffs.quantile(0.75)
    IQR = Q3 - Q1
    time_diffs_clean = time_diffs[(time_diffs >= Q1 - 1.5*IQR) & 
                                   (time_diffs <= Q3 + 1.5*IQR)]
    
    avg_interval = time_diffs_clean.mean()
    median_interval = time_diffs_clean.median()
    
    print(f"\nTime interval statistics (User {test_user}):")
    print(f"  Mean:   {avg_interval:.2f}")
    print(f"  Median: {median_interval:.2f}")
    
    # Determine sampling rate based on interval magnitude
    if avg_interval > 100:  # Likely milliseconds
        sampling_rate = 1000.0 / avg_interval
        print(f"  Unit: milliseconds")
    elif avg_interval > 1:  # Likely seconds
        sampling_rate = 1.0 / avg_interval
        print(f"  Unit: seconds")
    else:  # Small values, might be seconds
        sampling_rate = 1.0 / avg_interval
        print(f"  Unit: seconds (assumed)")
    
    print(f"\n✅ CALCULATED SAMPLING RATE: {sampling_rate:.2f} Hz")
    
    # Verify with multiple users
    print("\nVerification with multiple users:")
    rates = []
    for user in df['User'].unique()[:5]:
        user_data = df[df['User'] == user].sort_values(timestamp_col).head(500)
        if len(user_data) > 10:
            diffs = user_data[timestamp_col].diff().dropna()
            diffs = diffs[diffs > 0]  # Remove zeros
            if len(diffs) > 0:
                avg_diff = diffs.median()
                if avg_diff > 100:
                    rate = 1000.0 / avg_diff
                else:
                    rate = 1.0 / avg_diff
                rates.append(rate)
                print(f"  User {user}: {rate:.1f} Hz")
    
    if rates:
        final_rate = np.median(rates)
        print(f"\n🎯 FINAL ESTIMATE: {final_rate:.1f} Hz (median across users)")
    else:
        final_rate = sampling_rate
        print(f"\n🎯 FINAL ESTIMATE: {final_rate:.1f} Hz")

# Method 2: Estimate from signal characteristics (for StressSense or if timestamps fail)
if DATASET_TYPE == 'StressSense' or (timestamp_col is None):
    print("\n--- Method 2: Estimate from Activity Durations ---")
    
    # Count samples per activity segment
    activity_segments = []
    current_activity = None
    current_count = 0
    
    for activity in df['Activity'].head(2000):  # Sample first 2000
        if activity == current_activity:
            current_count += 1
        else:
            if current_count > 0:
                activity_segments.append(current_count)
            current_activity = activity
            current_count = 1
    
    if activity_segments:
        mean_segment = np.mean(activity_segments)
        median_segment = np.median(activity_segments)
        
        print(f"\nActivity segment statistics:")
        print(f"  Mean samples per segment: {mean_segment:.0f}")
        print(f"  Median samples per segment: {median_segment:.0f}")
        
        # Assume activities are 2-5 seconds long
        print("\nAssuming typical activity duration is 2-5 seconds:")
        common_rates = [20, 50, 100, 200]
        
        for rate in common_rates:
            duration = mean_segment / rate
            print(f"  {rate:>3d} Hz → {duration:.1f} sec/segment")
        
        # Best estimate (closest to 2-5 sec range)
        best_rate = 50  # default
        for rate in common_rates:
            duration = mean_segment / rate
            if 2 <= duration <= 10:
                best_rate = rate
                break
        
        print(f"\n🎯 ESTIMATED SAMPLING RATE: {best_rate} Hz")
        print(f"   (Based on {mean_segment:.0f} samples ≈ {mean_segment/best_rate:.1f} sec)")
        final_rate = best_rate

# Display recommended window sizes
print("\n" + "="*70)
print("RECOMMENDED WINDOW SIZES")
print("="*70)

if 'final_rate' in locals():
    print(f"\nUsing {final_rate:.0f} Hz sampling rate:")
    for duration in [2.0, 2.5, 3.0]:
        samples = int(final_rate * duration)
        # Round to nearest power of 2
        power_of_2 = 2 ** round(np.log2(samples))
        print(f"  {duration}s window = {samples} samples ≈ {power_of_2} samples (power of 2)")
    
    # Recommended
    recommended = 2 ** round(np.log2(final_rate * 2.5))
    print(f"\n✅ RECOMMENDED: {recommended} samples (~2.5 seconds)")
else:
    print("\n⚠️  Could not estimate sampling rate")
    print("   Using default: 128 samples (works for 20-50 Hz)")

print("="*70)
```

## Step 9: Visualize Activity Patterns

```python
print("="*70)
print("ACTIVITY SENSOR PATTERNS")
print("="*70)

# Select up to 4 activities
unique_activities = df['Activity'].unique()[:4]
n_activities = len(unique_activities)

print(f"\nVisualizing {n_activities} activities...")

fig, axes = plt.subplots(n_activities, 1, figsize=(14, n_activities * 2.5))
if n_activities == 1:
    axes = [axes]

# Use appropriate sensor column names
ax_col = sensor_cols[0]  # First accelerometer column
gx_col = sensor_cols[3]  # First gyroscope column

for i, activity in enumerate(unique_activities):
    activity_data = df[df['Activity'] == activity].head(200)
    
    axes[i].plot(activity_data[ax_col].values, label=f'{ax_col}', alpha=0.8, linewidth=1.5)
    axes[i].plot(activity_data[gx_col].values, label=f'{gx_col}', alpha=0.8, linewidth=1.5)
    
    axes[i].set_ylabel(activity, fontsize=11, weight='bold')
    axes[i].legend(loc='upper right')
    axes[i].grid(True, alpha=0.3)

axes[-1].set_xlabel('Sample Index', fontsize=11)
plt.suptitle('Sensor Patterns for Different Activities', fontsize=14, weight='bold')
plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}activity_patterns.png', dpi=300, bbox_inches='tight')
print(f"\n✓ Saved: {OUTPUT_DIR}activity_patterns.png")
plt.show()
```
